# Handling Missing Values

- 'age_certification' has a large number of missing values, so it will be filled with 'Unknown'.
- 'seasons' is only applicable for TV Shows, so missing values will be filled with 0 for Movies.
- Numerical columns like 'imdb_score', 'tmdb_score', and 'tmdb_popularity' will be filled using median values to avoid skewness.
- Missing 'description' and 'imdb_id' will not significantly impact analysis and can be left as is or dropped if required.

```python
# 1. Age certification
titles['age_certification'].fillna('Unknown', inplace=True)

# 2. Seasons (0 for movies)
titles['seasons'].fillna(0, inplace=True)

# 3. Fill numerical columns with median
titles['imdb_score'].fillna(titles['imdb_score'].median(), inplace=True)
titles['tmdb_score'].fillna(titles['tmdb_score'].median(), inplace=True)
titles['tmdb_popularity'].fillna(titles['tmdb_popularity'].median(), inplace=True)

# 4. Optional (less important columns)
titles['imdb_votes'].fillna(titles['imdb_votes'].median(), inplace=True)
```

#Business Question
What combination of factors leads to HIGH-RATED content?

```python
high_rated = titles_exploded[titles_exploded['imdb_score'] >= 7]

analysis = high_rated.groupby(
    ['type', 'genres', 'age_certification']
).agg({
    'imdb_score': 'mean',
    'runtime': 'mean',
    'release_year': 'count'
}).reset_index()

analysis.rename(columns={'release_year': 'content_count'}, inplace=True)

analysis_sorted = analysis.sort_values(by='imdb_score', ascending=False).head(15)

print(analysis_sorted)
```

```python
fig = px.scatter(
    analysis_sorted,
    x='runtime',
    y='imdb_score',
    size='content_count',
    color='type',
    hover_data=['genres', 'age_certification'],
    title="High Rated Content Pattern Analysis"
)

fig.show()
```

# Observation
High-rated content mostly:

Drama / Crime / Documentary
TV Shows dominate
Runtime ~90–120 mins (or episodic equivalent)
Mature certification (TV-MA, R

# Insight: High Performing Content Strategy

High-rated content is typically characterized by a combination of drama-oriented genres,
moderate runtime, mature age certifications, and is often produced as TV shows.
